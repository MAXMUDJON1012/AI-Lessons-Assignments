### Topshiriq 29
0 va 1 lardan iborat 5 x 5 checkerboard matritsa yarating.


```python
# Kodingizni shu yerga yozing
checkerboard = np.zeros((5, 5), dtype=int)
checkerboard[1::2, ::2] = 1
checkerboard[::2, 1::2] = 1

print(checkerboard)
```

### Topshiriq 30
6 x 6 matritsa yarating va chegara elementlarini 1, ichki elementlarini 0 qiling.


```python
# Kodingizni shu yerga yozing
matrix = np.zeros((6, 6), dtype=int)
matrix[0, :] = 1
matrix[-1, :] = 1
matrix[:, 0] = 1
matrix[:, -1] = 1

print(matrix)
```
