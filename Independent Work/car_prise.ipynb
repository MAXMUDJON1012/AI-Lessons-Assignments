```python
# Asl datani copy qilib olish
df_original = data.copy()   # asl data — tegilmaydi
df_clean = data.copy()      # shu bilan ishlaymiz
```

```python
print("Tozalashdan oldin:", df_clean.shape)
```

```python
# 1. YEAR — to'g'rilash
year_median = int(df_clean[(df_clean['year'] >= 1990) & (df_clean['year'] <= 2025)]['year'].median())
bad_year_mask = (df_clean['year'] < 1990) | (df_clean['year'] > 2025)
df_clean.loc[bad_year_mask, 'year'] = year_median
print(f"✅ year — {bad_year_mask.sum()} ta xato → {year_median} bilan almashtirildi")
```

```python
# 2. MILEAGE_KM — to'g'rilash
mileage_median = int(df_clean[(df_clean['mileage_km'] >= 0) & (df_clean['mileage_km'] <= 500000)]['mileage_km'].median())
bad_mileage_mask = (df_clean['mileage_km'] < 0) | (df_clean['mileage_km'] > 500000)
df_clean.loc[bad_mileage_mask, 'mileage_km'] = mileage_median
print(f"✅ mileage_km — {bad_mileage_mask.sum()} ta xato → {mileage_median} bilan almashtirildi")
```

```python
#  4. DATA CLEANING AND HANDLING MISSING DATA

# Demo uchun sun'iy NaN qiymatlar qo'shamiz
df_demo = df_clean.copy()
df_demo.loc[[5, 10, 20], 'price'] = np.nan
df_demo.loc[[3, 7], 'mileage_km'] = np.nan
df_demo.loc[[1], 'brand'] = np.nan

```

```python
# ── 1. isnull() — qaysi qiymatlar NaN? ───────────────
print("=== df.isnull().sum() ===")
print(df_demo.isnull().sum())

print("\n=== df.isnull() — birinchi 5 qator ===")
print(df_demo.isnull().head())

```

```python
# ── 2. notnull() — qaysi qiymatlar to'liq? ───────────
print("\n=== df.notnull().sum() ===")
print(df_demo.notnull().sum())

```

```python
# ── 3. dropna() — NaN qatorlarni o'chirish ────────────
print("\n=== df.dropna() — barcha NaN qatorlarni o'chirish ===")
print("Oldin:", df_demo.shape)
print("Keyin:", df_demo.dropna().shape)

```

```python
# Faqat 'price' ustunidagi NaN qatorlarni o'chirish
print("\n=== df.dropna(subset=['price']) ===")
print("Keyin:", df_demo.dropna(subset=['price']).shape)

```

```python
# Kamida 8 ta to'liq qiymatga ega qatorlarni saqlash
print("\n=== df.dropna(thresh=8) ===")
print("Keyin:", df_demo.dropna(thresh=8).shape)

```

```python
# axis=1 — NaN ustunlarni o'chirish
print("\n=== df.dropna(axis=1) — NaN ustunlarni o'chirish ===")
print("Keyin:", df_demo.dropna(axis=1).shape)

```

```python
# ── 4. fillna() — NaN ni to'ldirish ──────────────────
print("\n=== df.fillna() ===")

# Raqamli ustunlarni median bilan to'ldirish
df_filled = df_demo.copy()
df_filled['price'] = df_filled['price'].fillna(df_filled['price'].median())
df_filled['mileage_km'] = df_filled['mileage_km'].fillna(df_filled['mileage_km'].median())

# Matnli ustunni eng ko'p uchraydigan qiymat bilan to'ldirish
df_filled['brand'] = df_filled['brand'].fillna(df_filled['brand'].mode()[0])

print("NaN qoldi:", df_filled.isnull().sum().sum())  # 0 bo'lishi kerak

```

```python
# ── 5. replace() — qiymatlarni almashtirish ──────────
print("\n=== df.replace() ===")

# Yoqilg'i turlarini qisqartirish
df_replaced = df_clean.copy()
df_replaced['fuel_type'] = df_replaced['fuel_type'].replace({
    'petrol'   : 'Benzin',
    'diesel'   : 'Dizel',
    'electric' : 'Elektr',
    'hybrid'   : 'Gibrid'
})
print(df_replaced['fuel_type'].value_counts())

```

```python
# ── 6. duplicated() — takroriy qatorlar ──────────────
print("\n=== df.duplicated() ===")
print("Takroriy qatorlar soni:", df_clean.duplicated().sum())
print("Takroriy qatorlar:")
print(df_clean[df_clean.duplicated()])

# ── 7. drop_duplicates() — takroriylarni o'chirish ───
print("\n=== df.drop_duplicates() ===")
print("Oldin:", df_clean.shape)
df_no_dup = df_clean.drop_duplicates()
print("Keyin:", df_no_dup.shape)

# Faqat brand va year bo'yicha tekshirish
df_no_dup2 = df_clean.drop_duplicates(subset=['brand', 'year'])
print("brand+year bo'yicha keyin:", df_no_dup2.shape)
```

```python
# 5. DATA MANIPULATION



# ── 1. sort_values() — qiymat bo'yicha saralash ───────
print("=== df.sort_values() ===")

# Narx bo'yicha o'sish tartibida
print(df_clean.sort_values('price').head())

# Narx bo'yicha kamayish tartibida
print(df_clean.sort_values('price', ascending=False).head())

# Bir nechta ustun bo'yicha saralash
print(df_clean.sort_values(['brand', 'price'], ascending=[True, False]).head(10))

```

```python
# ── 2. sort_index() — index bo'yicha saralash ─────────
print("\n=== df.sort_index() ===")

# Avval indexni aralashtiramiz, keyin qayta tartiblaymiz
df_shuffled = df_clean.sample(frac=1, random_state=42)
print("Aralashgan index:", df_shuffled.index[:5].tolist())
print("Tartiblangan index:", df_shuffled.sort_index().index[:5].tolist())

```

```python
# ── 3. rename() — ustun nomlarini o'zgartirish ────────
print("\n=== df.rename() ===")
df_renamed = df_clean.rename(columns={
    'brand'       : 'marka',
    'year'        : 'yil',
    'mileage_km'  : 'yurgan_km',
    'fuel_type'   : 'yoqilgi',
    'transmission': 'uzatma',
    'engine_size' : 'dvigatel',
    'condition'   : 'holat',
    'price'       : 'narx'
})
print(df_renamed.columns.tolist())

```

```python
# ── 4. set_index() — ustunni index qilib belgilash ────
print("\n=== df.set_index() ===")
df_indexed = df_clean.set_index('car_id')
print(df_indexed.head())
print("Yangi index:", df_indexed.index[:5].tolist())

```

```python
# ── 5. reset_index() — indexni qayta 0 dan boshlash ──
print("\n=== df.reset_index() ===")
df_reset = df_indexed.reset_index()
print(df_reset.head())
print("Index qaytdi:", df_reset.index[:5].tolist())

```

```python
# ── 6. drop() — ustun yoki qatorlarni o'chirish ───────
print("\n=== df.drop() ===")

# Ustun o'chirish (axis=1)
df_dropped_col = df_clean.drop(columns=['car_id'])
print("car_id o'chirildi:", df_dropped_col.columns.tolist())

# Bir nechta ustun o'chirish
df_dropped_cols = df_clean.drop(columns=['car_id', 'engine_size'])
print("Ikki ustun o'chirildi:", df_dropped_cols.columns.tolist())

# Qator o'chirish (axis=0)
df_dropped_row = df_clean.drop(index=[0, 1, 2])
print("0,1,2 qatorlar o'chirildi, yangi boshi:", df_dropped_row.index[:3].tolist())

```

```python
# ── 7. insert() — yangi ustun qo'shish (aniq joyga) ──
print("\n=== df.insert() ===")
df_inserted = df_clean.copy()

# 3-pozitsiyaga 'age' ustunini qo'shish
df_inserted.insert(3, 'car_age', 2025 - df_inserted['year'])
print(df_inserted.columns.tolist())
print(df_inserted[['brand', 'year', 'car_age']].head())

```

```python
# ── 8. assign() — yangi ustun qo'shish (zanjirli) ────
print("\n=== df.assign() ===")
df_assigned = df_clean.assign(
    car_age       = 2025 - df_clean['year'],
    price_per_km  = (df_clean['price'] / df_clean['mileage_km']).round(2),
    is_expensive  = df_clean['price'] > df_clean['price'].median()
)
print(df_assigned[['brand', 'price', 'car_age', 'price_per_km', 'is_expensive']].head())
```

```python
#  6. AGGREGATION AND STATISTICS

# ── 1. sum() — yig'indi ───────────────────────────────
print("=== df.sum() ===")
print(df_clean[['price', 'mileage_km']].sum())

```

```python
# ── 2. mean() — o'rtacha ──────────────────────────────
print("\n=== df.mean() ===")
print(df_clean[['price', 'mileage_km', 'engine_size']].mean().round(2))

# Brand bo'yicha o'rtacha narx
print("\nBrand bo'yicha o'rtacha narx:")
print(df_clean.groupby('brand')['price'].mean().round(2).sort_values(ascending=False))
```

```python
# ── 3. median() — median ──────────────────────────────
print("\n=== df.median() ===")
print(df_clean[['price', 'mileage_km', 'engine_size']].median())

```

```python
# ── 4. min() — minimum ────────────────────────────────
print("\n=== df.min() ===")
print(df_clean[['price', 'mileage_km', 'year', 'engine_size']].min())

```

```python
# ── 5. max() — maksimum ───────────────────────────────
print("\n=== df.max() ===")
print(df_clean[['price', 'mileage_km', 'year', 'engine_size']].max())

```

```python
# ── 6. std() — standart og'ish ────────────────────────
print("\n=== df.std() ===")
print(df_clean[['price', 'mileage_km', 'engine_size']].std().round(2))

```

```python
# ── 7. var() — dispersiya ─────────────────────────────
print("\n=== df.var() ===")
print(df_clean[['price', 'mileage_km', 'engine_size']].var().round(2))

```

```python
# ── 8. count() — qiymatlar soni ──────────────────────
print("\n=== df.count() ===")
print(df_clean.count())

```

```python
# ── 9. corr() — korrelyatsiya ─────────────────────────
print("\n=== df.corr() ===")
corr = df_clean[['price', 'mileage_km', 'year', 'engine_size']].corr().round(3)
print(corr)
print("\nNarx bilan eng kuchli bog'liq:", corr['price'].drop('price').abs().idxmax())

```

```python
# ── 10. groupby() — guruhlab hisoblash ───────────────
print("\n=== df.groupby() ===")
# Brand bo'yicha guruhlash
print("\nBrand bo'yicha o'rtacha narx:")
print(df_clean.groupby('brand')['price'].mean().round(2))

# Fuel type bo'yicha guruhlash
print("\nYoqilg'i turi bo'yicha o'rtacha narx:")
print(df_clean.groupby('fuel_type')['price'].mean().round(2))

# Bir nechta ustun bo'yicha guruhlash
print("\nBrand + Transmission bo'yicha o'rtacha narx:")
print(df_clean.groupby(['brand', 'transmission'])['price'].mean().round(2))

# ── 11. agg() — bir nechta funksiya birga ────────────
print("\n=== df.agg() ===")

# Bitta ustunda bir nechta hisob
print("\n'price' ustuni bo'yicha:")
print(df_clean['price'].agg(['min', 'max', 'mean', 'median', 'std']).round(2))

# Har ustunda turli hisob
print("\nHar ustun uchun turli funksiya:")
print(df_clean[['price', 'mileage_km', 'engine_size']].agg({
    'price'      : ['min', 'max', 'mean', 'median'],
    'mileage_km' : ['min', 'max', 'mean'],
    'engine_size': ['min', 'max', 'mean']
}).round(2))

# groupby + agg — eng kuchli kombinatsiya
print("\nBrand bo'yicha to'liq statistika:")
print(df_clean.groupby('brand')['price'].agg(
    eng_arzon = 'min',
    eng_qimmat = 'max',
    ortacha   = 'mean',
    soni      = 'count'
).round(2))
```
